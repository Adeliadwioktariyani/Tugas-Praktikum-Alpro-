# Smart Vechile Insurace Analyzer

```python
# Input data kendaraan
usia_pengemudi = int(input("Masukkan usia pengemudi (tahun): "))
jenis_kendaraan = input("Masukkan jenis kendaraan (mobil/motor): ")
nilai_kendaraan = float(input("Masukkan nilai kendaraan (Rp): "))
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ")
sistem_pengaman = input("Apakah memiliki sistem pengaman? (ya/tidak): ")

# Menentukan risiko awal berdasarkan usia
if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = 3
else:
    risiko = 1

# Menentukan risiko berdasarkan jenis kendaraan
if usia_pengemudi >= 20 and usia_pengemudi <= 60:
    if jenis_kendaraan == "motor":
        risiko = 2
    elif jenis_kendaraan == "mobil":
        risiko = 1

# Risiko meningkat jika digunakan di perkotaan
if area_penggunaan == "perkotaan":
    risiko = risiko + 1

# Risiko meningkat jika tidak memiliki sistem pengaman 
if sistem_pengaman == "tidak":
    risiko = risiko + 1

# Risiko maksimal adalah 3 (tinggi)
if risiko > 3:
    risiko = 3

# Mengubah angka menjadi kategori risiko
if risiko == 3:
    kategori_risiko = "tinggi"
elif risiko == 2:
    kategori_risiko = "sedang"
else:
    kategori_risiko = "rendah"

# Menentukan persentase premi
if kategori_risiko == "tinggi":
    persentase_premi = 0.10
elif kategori_risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

# Menghitung premi asuransi
premi = persentase_premi * nilai_kendaraan

# Pesan doraemon menggunakan conditional expression
pesan = (
    "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if kategori_risiko == "tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon."
)

# Menampilkan hasil
print("\n=== HASIL ANALISIS ASURANSI DORABIKE ===")
print("Kategori risiko:", kategori_risiko)
print("Persentase premi:", persentase_premi * 100, "%")
print("premi asuransi: Rp", format(premi, ",.0f"))
print("Pesan Doraemon:", pesan)
```

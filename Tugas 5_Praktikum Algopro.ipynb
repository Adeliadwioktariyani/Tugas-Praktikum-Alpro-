```python
# A. Mencatat pengeluaran harian Nobita
hari = 1
saldo = 50000

while hari <= 30:
    pengeluaran= int(input(f"Pengeluaran hari ke-{hari}: "))
    saldo -= pengeluaran
    if saldo < 0:
        print(f"Saldo sudah habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break
    print(f"Sisa saldo hari ke-{hari}: Rp{saldo}")
    hari += 1
else:
    print("Satu bulan selesai, saldo Nobita masih aman!")
    
```

```python
# B. Menganalisis Pengeluaran Mingguan Nobita
pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in range(4):
    if pengeluaran_mingguan[i] == 0:
        print(f"Minggu ke-{i + 1}: Nobita tidak jajan, dilewati analisisnya.")
        continue
    print(f"Minggu ke-{i + 1}: Total pengeluaran Rp{pengeluaran_mingguan[i]}")
else:
    print("Analisis mingguan selesai.")
```

```python
# C. Menyusun Laporan Bulanan Nobita
total_semua = 0
bulan_dihentikan = 0

for bulan in range(1, 4):
    print(f"Bulan ke-{bulan}")
    total_bulan = 0

    for minggu in range(1, 5):
        pengeluaran = int(input(f"Pengeluaran minggu ke-{minggu}: "))
        total_bulan += pengeluaran 
        
        if pengeluaran > 20000:
            print("Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
            bulan_dihentikan += 1
            break
    else:
        print(f"total pengeluaran bulan ke-{bulan}: Rp{total_bulan}")

    total_semua += total_bulan

print("\n === Ringkasan 3 Bulan ===")
print(f"Total seluruh pengeluaran tercatat: Rp{total_semua}")
print(f"Jumlah bulan yang dihentikan: {bulan_dihentikan}")
```

```python
# D. Evaluasi Akhir Tahun
saldo_awal = 300000
saldo = saldo_awal

for bulan in range(1, 13):
    print (f"=== Bulan ke-{bulan} ===")
    minggu = 1

    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur? (y/n): ")

        if libur == "y":
            print("Minggu ini dilewati (continue).")
            minggu += 1  
            continue

        pengeluaran = int(input("Masukkan pengeluaran minggu ini: "))
        saldo -= pengeluaran
        minggu += 1

        if saldo < 0:
            break
    else:
        print(f"Saldo tersisa akhir bulan {bulan}: Rp{saldo}")

    if saldo < 0:
        print(f"Saldo Nobita habis di bulan ke-{bulan}. Simulasi dihentikan!")
        break
else:
    print(f"Nobita berhasil menabung Rp{saldo} dari total Rp{saldo_awal}!")  
```

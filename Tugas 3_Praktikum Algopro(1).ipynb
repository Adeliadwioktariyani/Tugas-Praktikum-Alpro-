```python
# Membuat Data Belanjaan
list_belanjaan = ["Apel", "Susu", "Roti"]
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

print(f"Karakter pertama dari barang pertama ialah ", list_belanjaan[0][0])
print(f"Karakter terakhir dari barang terakhir ialah ", list_belanjaan[-1][-1])
```

```python
# Hitung Total Harga per Barang
total_apel = harga_jumlah[0] * harga_jumlah[1] # 5000 * 3 = 15000
total_susu = harga_jumlah[2] * harga_jumlah[3] # 12000 * 2 = 24000
total_roti = harga_jumlah[4] * harga_jumlah[5] # 8000 * 1 = 8000

total_belanja = total_apel + total_susu + total_roti  # 47000
```

```python
# Cetak Struk Belanja
nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Sembako Doraemon"

print('=========================================')
print('       Toko Sembako Doraemon             ')
print('=========================================')
print(f"Pembeli: {nama_pembeli}\n")
print(f"Barang\t\tHarga\tJumlah\tTotal")
print('-----------------------------------------')
print(f"{list_belanjaan[0]}\t\t{harga_jumlah[0]}\t{harga_jumlah[1]}\t{total_apel}")
print(f"{list_belanjaan[1]}\t\t{harga_jumlah[2]}\t{harga_jumlah[3]}\t{total_susu}")
print(f"{list_belanjaan[2]}\t\t{harga_jumlah[4]}\t{harga_jumlah[5]}\t{total_roti}")
print('-----------------------------------------')
print(f"Total belanja\t\t\t{total_belanja}")
print('=========================================')
print('    Terima kasih sudah belanja! 🙏🏻      ')
print('=========================================')
```
